## 2. Una tabla de datos

`pandas` es la librería que usaremos para trabajar con tablas, como una hoja de cálculo.

```python
import pandas as pd

pacientes = pd.DataFrame({
    "edad": [34, 58, 71, 45, 62],
    "presion_sistolica": [118, 142, 155, 126, 138],
    "diabetes": ["no", "sí", "sí", "no", "no"],
})
pacientes
```

```python
pacientes["presion_sistolica"].mean()
```

## 3. Un primer gráfico

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 3.5))
ax.scatter(pacientes["edad"], pacientes["presion_sistolica"])
ax.set_xlabel("Edad (años)")
ax.set_ylabel("Presión sistólica (mmHg)")
plt.show()
```
